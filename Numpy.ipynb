```python
import numpy as np

#TASK 1
array = np.array([10, 20, 30, 40, 50])
print ("Sum:",np.sum(array))
print ("Average:",np.mean(array))
print ("Maximum:",np.max(array))
print ("Minimum:",np.min(array))
```

```python
#TASK 2
array = np.array([ 45, 80, 30, 90, 55, 20])
print (array[array >= 50 ] )
```

```python
#TASK 4
marks = np.random.randint(40, 100, 10)
print ("Marks:",marks)


print ("Average:",np.mean(marks))
print ("Highest marks:",np.max(marks))
print ("Lowest marks:",np.min(marks))
```

```python
#TASK 5

masks= np.array([35, 67, 89, 45, 90, 22, 70, 58])

passed = masks[masks >= 50 ]
failed = masks[masks < 50 ]

print ("Passed marks:")
print (passed)
print ("\nfailed marks:")
print (failed)

print ("\nNumber of passed students:", len(passed))
print ("Number of passed students:", len(failed))
```
