```python
#Task 11: Personal Expense Tracker

expenses = []

total_expenses = int(input("Enter total expenses: "))

for i in range(total_expenses):
    name = input("Enter expense name: ")
    category = input("Enter category: ")
    amount = float(input("Enter amount: "))

    expense = {
        "name": name,
        "category": category,
        "amount": amount
    }

    expenses.append(expense)

budget = float(input("Enter budget: "))

total = 0
category_total = {}

for expense in expenses:
    total = total + expense["amount"]

    category = expense["category"]

    if category in category_total:
        category_total[category] = category_total[category] + expense["amount"]
    else:
        category_total[category] = expense["amount"]

highest = expenses[0]

for expense in expenses:
    if expense["amount"] > highest["amount"]:
        highest = expense

print("Total Spending:", total)

for category, amount in category_total.items():
    print(category, ":", amount)

print("Highest Expense:", highest["name"])
print("Highest Amount:", highest["amount"])

if total > budget:
    print("Warning: Budget Exceeded")
else:
    print("Budget Remaining:", budget - total)
```

```python
#Task 12: Store Inventory Manager

products = {}

def add_product():
    name = input("Enter product name: ")
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))

    if name in products:
        print("Product already exists")
    else:
        products[name] = {
            "price": price,
            "quantity": quantity
        }
        print("Product added")

def sell_product():
    name = input("Enter product name: ")
    quantity = int(input("Enter quantity to sell: "))

    if name not in products:
        print("Product not found")
    elif products[name]["quantity"] < quantity:
        print("Insufficient stock")
    else:
        products[name]["quantity"] -= quantity
        print("Product sold")

def restock_product():
    name = input("Enter product name: ")
    quantity = int(input("Enter quantity to add: "))

    if name in products:
        products[name]["quantity"] += quantity
        print("Product restocked")
    else:
        print("Product not found")

def search_product():
    name = input("Enter product name: ")

    if name in products:
        print(products[name])
    else:
        print("Product not found")

def display_inventory():
    total_value = 0

    for name, product in products.items():
        print(name, product)

        value = product["price"] * product["quantity"]
        total_value += value

        if product["quantity"] < 5:
            print("Low Stock:", name)

    print("Total Inventory Value:", total_value)

while True:
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_product()
    elif choice == "2":
        sell_product()
    elif choice == "3":
        restock_product()
    elif choice == "4":
        search_product()
    elif choice == "5":
        display_inventory()
    elif choice == "6":
        break
    else:
        print("Invalid choice")
```

```python
#Task 13: Library Book Management System

books = []

def add_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists")
            return

    title = input("Enter title: ")
    author = input("Enter author: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)
    print("Book added")

def display_books():
    for book in books:
        print(book)

def search_book():
    title = input("Enter title: ")

    for book in books:
        if book["title"].lower() == title.lower():
            print(book)
            return

    print("Book not found")

def borrow_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            if book["available"]:
                book["available"] = False
                print("Book borrowed")
            else:
                print("Book is already borrowed")
            return

    print("Book not found")

def return_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            book["available"] = True
            print("Book returned")
            return

    print("Book not found")

while True:
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        borrow_book()
    elif choice == "5":
        return_book()
    elif choice == "6":
        break
    else:
        print("Invalid choice")
```

```python
#Task 14: Restaurant Ordering System

menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

orders = []

def display_menu():
    for item, price in menu.items():
        print(item, ":", price)

def take_order():
    item = input("Enter item name: ")

    if item in menu:
        quantity = int(input("Enter quantity: "))

        order = {
            "item": item,
            "quantity": quantity,
            "subtotal": menu[item] * quantity
        }

        orders.append(order)
        print("Item added")
    else:
        print("Item not available")

def receipt():
    total = 0

    for order in orders:
        print(order["item"], "x", order["quantity"], "=", order["subtotal"])
        total += order["subtotal"]

    if total >= 3000:
        discount = total * 10 / 100
    else:
        discount = 0

    after_discount = total - discount
    tax = after_discount * 13 / 100
    final_bill = after_discount + tax

    print("Total:", total)
    print("Discount:", discount)
    print("Tax:", tax)
    print("Final Bill:", final_bill)

display_menu()

while True:
    take_order()

    more = input("Add another item? yes/no: ")

    if more.lower() == "no":
        break

receipt()
```

```python
#Task 15: Clinic Appointment Manager

patients = []

def add_appointment():
    patient_id = input("Enter patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists")
            return

    name = input("Enter patient name: ")
    age = int(input("Enter age: "))
    disease = input("Enter disease/problem: ")
    time = input("Enter appointment time: ")
    emergency = input("Emergency? yes/no: ")

    patient = {
        "id": patient_id,
        "name": name,
        "age": age,
        "disease": disease,
        "time": time,
        "emergency": emergency.lower() == "yes"
    }

    patients.append(patient)
    print("Appointment added")

def search_patient():
    patient_id = input("Enter patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print(patient)
            return

    print("Patient not found")

def display_appointments():
    for patient in patients:
        if patient["emergency"]:
            print(patient)

    for patient in patients:
        if not patient["emergency"]:
            print(patient)

def cancel_appointment():
    patient_id = input("Enter patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Appointment cancelled")
            return

    print("Patient not found")

def clinic_summary():
    emergency_count = 0
    normal_count = 0

    for patient in patients:
        if patient["emergency"]:
            emergency_count += 1
        else:
            normal_count += 1

    print("Emergency Patients:", emergency_count)
    print("Normal Patients:", normal_count)
    print("Total Patients:", len(patients))

while True:
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_appointment()
    elif choice == "2":
        search_patient()
    elif choice == "3":
        display_appointments()
    elif choice == "4":
        cancel_appointment()
    elif choice == "5":
        clinic_summary()
    elif choice == "6":
        break
    else:
        print("Invalid choice")
```
