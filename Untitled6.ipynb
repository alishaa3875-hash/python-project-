```python
import pandas as pd
#TASK1
data =  {
    "Name": ["Ali","Sara","Ahmed","Ayesha","Zain"],
    "Marks":[70,85,60,95,45]
}
df = pd.DataFrame(data)
print(df)
print("\nAverage Marks:")
print(df["Marks"].mean())
print("\nMaximum marks:")
print(df["Marks"].max())
print("\nMinimum marks:")
print(df["Marks"].min())
print("\nTotal students:",len(df))
```

```python
#Task 2
data =  {
    "Name": ["Ali","Sara","Ahmed","Ayesha","Zain"],
    "Marks":[70,85,60,95,45]
}
df = pd.DataFrame(data)

print("\nPassed student:")
passed = df[df["Marks"] >= 50]
print(passed)

print("\nfailed student:")
failed = df[df["Marks"] < 50]
print(failed)
```

```python
#Task 3
data =  {
    "Name": ["Ali","Sara","Ahmed","Ayesha","Zain"],
    "Marks":[70,85,60,95,45]
}
df = pd.DataFrame(data)
df["City"] = ["Karachi", "Lahore", "Karachi", "Islamabad", "Karachi"]

karachi_students = df[df["City"] == "Karachi"]

print(karachi_students)
```

```python
#Task 4
data =  {
    "Name": ["Ali","Sara","Ahmed","Ayesha","Zain"],
    "Course":["AI","Web","AI","Web","AI"],
    "Marks":[70,85,60,95,45]
}
df = pd.DataFrame(data)
result = df.groupby("Course")["Marks"].mean()

print(result)
```

```python
#Task 5
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
   "Marks": [80, None, 70, None, 90]
}

df = pd.DataFrame(data)

# Number of missing values
print("Missing Values:", df["Marks"].isna().sum())

# Average marks
average = df["Marks"].mean()

# Fill missing values
df["Marks"] = df["Marks"].fillna(average)

print(df)
```
